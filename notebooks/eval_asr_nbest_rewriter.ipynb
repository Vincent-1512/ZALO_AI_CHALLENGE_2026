# Đánh Giá Chất Lượng Phiên Âm ASR PhoWhisper: N-Best Hypotheses & LLM Rewriter (1.5B)

## Mục Tiêu Thực Nghiệm:
1. **Khai thác $N$-best Hypotheses từ PhoWhisper:** Thay vì chỉ lấy 1 kết quả greedy dễ nhầm ngữ âm, sử dụng Beam Search để trích xuất $N$ câu dự đoán kèm điểm số tự tin (`log_prob`).
2. **LLM Rewriter (1.5B) Suy Luận Dựa Trên Lịch Sử:** Sử dụng mô hình `DeepSeek-R1-Distill-Qwen-1.5B` (kích thước vừa vặn trong ngân sách $\le 2.1\text{B}$ còn dư của cuộc thi) đọc $N$-best hypotheses kết hợp `history` từ `train.jsonl` để sửa lỗi chính tả, chuẩn hóa số tiền và giải quyết thực thể tên riêng/ngày giờ.
3. **Đo lường định lượng:** So sánh WER (Word Error Rate), CER (Character Error Rate) trước vs. sau khi có LLM Rewriter.

## 2. Xây Dựng Hàm Trích Xuất N-Best Hypotheses Từ PhoWhisper

Mô hình `vinai/PhoWhisper-medium` (~769M tham số) được cấu hình chạy Beam Search ($N=3$ hoặc $N=5$) và xuất ra chuỗi văn bản kèm điểm số tự tin chuẩn hóa.

```python
from transformers import AutoModelForSpeechSeq2Seq, AutoProcessor

ASR_MODEL_ID = "vinai/PhoWhisper-medium"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
DTYPE = torch.float16 if DEVICE == "cuda" else torch.float32

print(f"Đang tải mô hình ASR: {ASR_MODEL_ID} trên {DEVICE}...")
asr_processor = AutoProcessor.from_pretrained(ASR_MODEL_ID)
asr_model = AutoModelForSpeechSeq2Seq.from_pretrained(
    ASR_MODEL_ID,
    torch_dtype=DTYPE,
    low_cpu_mem_usage=True,
).to(DEVICE).eval()
print("Tải PhoWhisper thành công!")

def read_and_resample_audio(audio_path: str, target_sr: int = 16000) -> np.ndarray:
    audio, rate = sf.read(str(audio_path), dtype="float32", always_2d=True)
    audio = audio.mean(axis=1)
    if rate != target_sr:
        ratio = Fraction(target_sr, rate)
        audio = resample_poly(audio, ratio.numerator, ratio.denominator).astype(np.float32)
    return audio

def transcribe_nbest(
    audio_path: str,
    num_beams: int = 3,
    num_return_sequences: int = 3,
) -> List[Dict[str, Any]]:
    """Trích xuất N giả thuyết kèm theo điểm tự tin từ PhoWhisper."""
    audio = read_and_resample_audio(audio_path)
    inputs = asr_processor(audio, sampling_rate=16000, return_tensors="pt")
    input_features = inputs.input_features.to(DEVICE, dtype=DTYPE)

    with torch.inference_mode():
        outputs = asr_model.generate(
            input_features,
            num_beams=num_beams,
            num_return_sequences=num_return_sequences,
            return_dict_in_generate=True,
            output_scores=True,
        )

    sequences = outputs.sequences
    # Tính normalized sequence scores
    scores = outputs.sequences_scores.cpu().numpy() if hasattr(outputs, "sequences_scores") else [0.0] * len(sequences)

    results = []
    for seq, score in zip(sequences, scores):
        text = asr_processor.batch_decode([seq], skip_special_tokens=True)[0]
        text = unicodedata.normalize("NFC", text).strip()
        # Chuyển log prob sang confidence xấp xỉ (0% -> 100%)
        prob_pct = round(math.exp(min(0.0, float(score))) * 100, 1)
        results.append({
            "text": text,
            "log_prob": float(score),
            "confidence": prob_pct
        })
    return results
```

```python
# Xác định đường dẫn file dữ liệu
    candidate_dirs = [
        Path("/content/drive/MyDrive/ZALO/ZALO_AI_CHALLENGE_2026/train+public_test/train"),
        Path("/content/drive/MyDrive/ZALO/ZALO_AI_CHALLENGE_2026/data/raw/train+public_test/train"),
        Path("/content/drive/MyDrive/ZALO/train+public_test/train"),
        Path("../train+public_test/train"),
        Path("train+public_test/train"),
        Path("data/raw/train+public_test/train"),
        Path("../data/raw/train+public_test/train"),
    ]
    DATA_DIR = next((p for p in candidate_dirs if p.exists() and (p / "train.jsonl").is_file()), candidate_dirs[0])

train_jsonl = DATA_DIR / "train.jsonl"
print("File dữ liệu train:", train_jsonl)

samples = []
with open(train_jsonl, "r", encoding="utf-8") as f:
    for line in f:
        if line.strip():
            samples.append(json.loads(line))

print(f"Tổng số mẫu trong train: {len(samples)}")
# Lấy 30 mẫu có audio để chạy thực nghiệm đánh giá
test_subset = [s for s in samples if s.get("audio") and (DATA_DIR / s["audio"]).exists()][:30]
if not test_subset:
    # Fallback tìm trong thư mục audio nếu đường dẫn tương đối khác
    test_subset = samples[:30]

print(f"Đã chuẩn bị {len(test_subset)} mẫu để test!")
print("Mẫu đầu tiên:", {"id": test_subset[0]["id"], "audio": test_subset[0].get("audio"), "audio_text": test_subset[0].get("audio_text")})
```

## 4. Tải Mô Hình LLM Rewriter (1.5B Tham Số)

Sử dụng `DeepSeek-R1-Distill-Qwen-1.5B` (hoặc `Qwen2.5-1.5B-Instruct` làm fallback) để thực hiện tư duy suy luận Chain-of-Thought.

```python
from transformers import AutoModelForCausalLM, AutoTokenizer

REWRITER_ID = "deepseek-ai/DeepSeek-R1-Distill-Qwen-1.5B"
print(f"Đang tải LLM Rewriter: {REWRITER_ID}...")

try:
    rewriter_tokenizer = AutoTokenizer.from_pretrained(REWRITER_ID, trust_remote_code=True)
    rewriter_model = AutoModelForCausalLM.from_pretrained(
        REWRITER_ID,
        torch_dtype=torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16,
        device_map="auto" if torch.cuda.is_available() else None,
        low_cpu_mem_usage=True,
        trust_remote_code=True,
    ).eval()
    print("Tải DeepSeek-R1-1.5B thành công!")
except Exception as e:
    print(f"Không tải được DeepSeek-R1 ({e}), chuyển sang fallback Qwen2.5-1.5B-Instruct...")
    REWRITER_ID = "Qwen/Qwen2.5-1.5B-Instruct"
    rewriter_tokenizer = AutoTokenizer.from_pretrained(REWRITER_ID)
    rewriter_model = AutoModelForCausalLM.from_pretrained(
        REWRITER_ID,
        torch_dtype=torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16,
        device_map="auto" if torch.cuda.is_available() else None,
    ).eval()
```

## 5. Thiết Kế Prompt Sửa Lỗi Ngữ Âm & Suy Luận Lịch Sử

Prompt cung cấp cho LLM:
1. **Danh sách $N$-best hypotheses** kèm độ tự tin.
2. **Lịch sử hội thoại `history`** (các câu người dùng nói trước đó, tên liên hệ, cuộc gọi, việc làm).
3. **Ngữ cảnh thời gian `current_time`**.

```python
REWRITER_PROMPT_TEMPLATE = """Bạn là chuyên gia hiệu đính tiếng Việt thông minh cho trợ lý ảo LifeMate.
Nhiệm vụ của bạn là xem xét các giả thuyết nhận dạng giọng nói từ mô hình ASR (kèm độ tự tin), đối chiếu với lịch sử hội thoại trước đó của người dùng và thời gian hiện tại để khôi phục lại câu lệnh gốc chuẩn xác nhất.

Ngữ cảnh:
- Thời gian hiện tại: {current_time}
- Lịch sử hội thoại trước đó:
{history_str}

Danh sách các giả thuyết ASR nhận diện được:
{hypotheses_str}

Hãy suy nghĩ trong thẻ <think>...</think>:
1. Phân tích các từ bị mâu thuẫn giữa các giả thuyết (đặc biệt là số tiền, tên người, từ đồng âm, từ địa phương).
2. Dựa vào lịch sử để tìm manh mối (ví dụ tên người đã từng liên lạc, tác vụ đang dở).
3. Chuẩn hóa lại ngày giờ nếu cần thiết.

Sau thẻ </think>, xuất đúng một dòng duy nhất theo định dạng:
[CORRECTED]: <câu_lệnh_tiếng_Việt_chuẩn_xác>"""

def format_history(history: List[Dict[str, Any]]) -> str:
    if not history:
        return "(Không có lịch sử trước đó)"
    lines = []
    for idx, turn in enumerate(history[-4:], 1):
        role = turn.get("role", "user")
        text = turn.get("text", "") or turn.get("audio_text", "")
        calls = turn.get("tool_calls", [])
        call_str = f" -> Gọi tool: {[c.get('name') for c in calls]}" if calls else ""
        lines.append(f"- Lượt {idx} ({role}): {text}{call_str}")
    return "\n".join(lines)

def rewrite_with_llm(hypotheses: List[Dict[str, Any]], history: List[Dict[str, Any]], current_time: str) -> str:
    hyp_str = "\n".join([
        f"  + Giả thuyết {idx+1} (tự tin {h['confidence']}%): \"{h['text']}\""
        for idx, h in enumerate(hypotheses)
    ])
    hist_str = format_history(history)
    prompt = REWRITER_PROMPT_TEMPLATE.format(
        current_time=current_time,
        history_str=hist_str,
        hypotheses_str=hyp_str,
    )

    messages = [{"role": "user", "content": prompt}]
    prompt_text = rewriter_tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = rewriter_tokenizer(prompt_text, return_tensors="pt").to(DEVICE)

    with torch.inference_mode():
        out = rewriter_model.generate(
            **inputs,
            max_new_tokens=300,
            do_sample=False,
            pad_token_id=rewriter_tokenizer.eos_token_id,
        )
    gen_text = rewriter_tokenizer.decode(out[0][inputs["input_ids"].shape[-1]:], skip_special_tokens=True)
    
    # Bóc tách [CORRECTED]:
    match = re.search(r"\[CORRECTED\]:\s*(.*)", gen_text, flags=re.IGNORECASE)
    if match:
        return unicodedata.normalize("NFC", match.group(1).strip())
    # Fallback nếu không có tag
    lines = [l.strip() for l in gen_text.splitlines() if l.strip() and not l.startswith("<")]
    return lines[-1] if lines else hypotheses[0]["text"]
```

## 6. Chạy Thực Nghiệm Định Lượng (WER / CER) So Sánh Trước vs. Sau

```python
import jiwer
from tqdm import tqdm

transform = jiwer.Compose([
    jiwer.ToLowerCase(),
    jiwer.RemovePunctuation(),
    jiwer.RemoveMultipleSpaces(),
    jiwer.Strip(),
])

def normalize_for_metric(text: str) -> str:
    return transform(unicodedata.normalize("NFC", text))

raw_asr_preds = []
rewritten_preds = []
ground_truths = []
cases = []

print("Bắt đầu chạy thử nghiệm trên các mẫu...")
for idx, sample in enumerate(tqdm(test_subset)):
    audio_rel = sample.get("audio", "")
    audio_path = DATA_DIR / audio_rel
    gt_text = sample.get("audio_text", "")
    if not audio_path.exists() or not gt_text:
        continue

    # 1. Trích xuất N-best từ PhoWhisper
    nbest = transcribe_nbest(str(audio_path), num_beams=3, num_return_sequences=3)
    raw_top1 = nbest[0]["text"]

    # 2. Đưa qua LLM Rewriter sửa lỗi
    ctx = sample.get("context", {})
    cur_time = ctx.get("current_time", "2026-11-20T16:00:00+07:00")
    hist = sample.get("history", [])
    corrected = rewrite_with_llm(nbest, hist, cur_time)

    ground_truths.append(normalize_for_metric(gt_text))
    raw_asr_preds.append(normalize_for_metric(raw_top1))
    rewritten_preds.append(normalize_for_metric(corrected))

    cases.append({
        "id": sample["id"],
        "gt": gt_text,
        "raw_top1": raw_top1,
        "nbest": [h["text"] for h in nbest],
        "corrected": corrected,
    })

# 3. Tính toán các chỉ số lỗi
raw_wer = jiwer.wer(ground_truths, raw_asr_preds) * 100
raw_cer = jiwer.cer(ground_truths, raw_asr_preds) * 100

rewritten_wer = jiwer.wer(ground_truths, rewritten_preds) * 100
rewritten_cer = jiwer.cer(ground_truths, rewritten_preds) * 100

wer_reduction = (raw_wer - rewritten_wer) / raw_wer * 100 if raw_wer > 0 else 0.0

print("\n================ KẾT QUẢ ĐÁNH GIÁ ĐỊNH LƯỢNG ================")
print(f"1. PhoWhisper 1-best thô ban đầu : WER = {raw_wer:.2f}% | CER = {raw_cer:.2f}%")
print(f"2. Sau LLM Rewriter (N-best + Hist): WER = {rewritten_wer:.2f}% | CER = {rewritten_cer:.2f}%")
print(f"==> TỶ LỆ GIẢM LỖI (Relative Error Reduction): {wer_reduction:+.2f}%")
print("=============================================================")
```

```python
metrics = ["Word Error Rate (WER)", "Character Error Rate (CER)"]
before = [raw_wer, raw_cer]
after = [rewritten_wer, rewritten_cer]

x = np.arange(len(metrics))
width = 0.35

fig, ax = plt.subplots(figsize=(8, 5))
rects1 = ax.bar(x - width/2, before, width, label="PhoWhisper Gốc (1-best)", color="salmon")
rects2 = ax.bar(x + width/2, after, width, label="Sau LLM Rewriter (N-best + History)", color="royalblue")

ax.set_ylabel("Tỷ Lệ Lỗi (%) - Càng thấp càng tốt")
ax.set_title("Hiệu Quả Sửa Lỗi Của LLM Rewriter (1.5B) Trên PhoWhisper")
ax.set_xticks(x)
ax.set_xticklabels(metrics)
ax.legend()
ax.grid(axis="y", linestyle="--", alpha=0.7)

for rect in rects1 + rects2:
    height = rect.get_height()
    ax.annotate(f"{height:.2f}%",
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3),
                textcoords="offset points",
                ha="center", va="bottom", fontweight="bold")

plt.tight_layout()
plt.savefig("asr_rewriter_comparison.png", dpi=150)
plt.show()
print("Đã lưu biểu đồ so sánh: asr_rewriter_comparison.png")
```
